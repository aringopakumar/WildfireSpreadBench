## 9. Train UTAE (temporal model)

U-TAE is a **temporal** model, so it needs the multitemporal data config (`cfgs/data_multitemporal_full_features.yaml`: 5 leading observations + day-of-year), **not** the monotemporal config used above.

`return_doy: true` is required — U-TAE feeds each observation's day-of-year into its temporal attention encoder. Running the UTAE config against the monotemporal data (`return_doy: false`) will crash when the model tries to unpack `doys` from the batch.

Everything else is the same as the U-Net run: checkpoints go to Drive, the `UnifiedEvalCallback` reports `eval/*` every 10 epochs, and `do_test=True` runs the full test set at the end.

> **Heads up on checkpoints:** `trainer_colab.yaml` writes `last.ckpt` to the same Drive folder for every model. If you've trained U-Net in this session, that `last.ckpt` is a U-Net — don't resume UTAE from it. Point `--trainer.callbacks.init_args.dirpath` at a UTAE-specific folder (or clear the folder) before a long UTAE run.


```python
import os, glob, time, shutil, yaml

LOCAL_DATA_DIR = "/content/wildfire_data_local"
_marker = os.path.join(LOCAL_DATA_DIR, ".copy_complete")
MIN_HDF5 = 100  # refuse to "finish" with a near-empty local copy


def _count_hdf5(root):
    return len(glob.glob(os.path.join(root, "**", "*.hdf5"), recursive=True))


def _remount_drive():
    """Recover from a dead Drive FUSE mount (ENOTCONN / empty DATA_DIR)."""
    import subprocess
    from google.colab import drive
    try:
        drive.flush_and_unmount()
    except Exception:
        pass
    for cmd in (
        ["fusermount", "-u", "/content/drive"],
        ["umount", "-l", "/content/drive"],
    ):
        subprocess.run(cmd, check=False, capture_output=True)
    time.sleep(2)
    # Dead FUSE can leave a non-empty mountpoint that blocks drive.mount()
    if os.path.isdir("/content/drive") and not os.path.ismount("/content/drive"):
        try:
            for name in os.listdir("/content/drive"):
                p = os.path.join("/content/drive", name)
                if os.path.isdir(p) and not os.path.islink(p):
                    shutil.rmtree(p, ignore_errors=True)
                else:
                    try:
                        os.remove(p)
                    except OSError:
                        pass
        except OSError:
            pass
    drive.mount("/content/drive", force_remount=True)
    time.sleep(2)
    assert os.path.isdir(DATA_DIR), f"Drive remounted but DATA_DIR missing: {DATA_DIR}"
    print("  [Drive remounted]")


def copy_tree_robust(src, dst):
    n_copied = n_skipped = 0
    errors = []
    src_files = []
    for root, _dirs, files in os.walk(src):
        for fn in files:
            src_files.append(os.path.join(root, fn))
    if not src_files:
        # Walk saw nothing — mount is dead, not "copy complete"
        return 0, 0, [("WALK", "os.walk returned 0 files — Drive mount is dead")]
    for s in src_files:
        rel = os.path.relpath(s, src)
        d = os.path.join(dst, rel)
        os.makedirs(os.path.dirname(d), exist_ok=True)
        try:
            ssz = os.path.getsize(s)
            if os.path.isfile(d) and os.path.getsize(d) == ssz:
                n_skipped += 1
                continue
            tmp = d + ".tmp"
            shutil.copyfile(s, tmp)
            os.replace(tmp, d)
            n_copied += 1
            if n_copied % 20 == 0:
                print(f"  copied {n_copied} files...")
        except Exception as e:
            errors.append((s, repr(e)))
    return n_copied, n_skipped, errors


# If a previous run wrote a marker with almost no data, wipe the marker and retry
if os.path.exists(_marker) and _count_hdf5(LOCAL_DATA_DIR) < MIN_HDF5:
    print(f"Local copy looks incomplete ({_count_hdf5(LOCAL_DATA_DIR)} hdf5). Removing marker and retrying.")
    os.remove(_marker)

if not os.path.exists(_marker):
    print(f"Copying dataset {DATA_DIR} -> {LOCAL_DATA_DIR} (one-time, be patient)...")
    os.makedirs(LOCAL_DATA_DIR, exist_ok=True)
    MAX_PASSES = 25
    errors = [("init", "start")]
    for attempt in range(1, MAX_PASSES + 1):
        if not os.path.isdir(DATA_DIR):
            print("  DATA_DIR missing; remounting Drive...")
            _remount_drive()
        n_copied, n_skipped, errors = copy_tree_robust(DATA_DIR, LOCAL_DATA_DIR)
        print(f"Pass {attempt}: copied {n_copied}, skipped {n_skipped}, errors {len(errors)}")
        if not errors and _count_hdf5(LOCAL_DATA_DIR) >= MIN_HDF5:
            break
        print("  Remounting Drive and retrying...")
        try:
            _remount_drive()
        except Exception as e:
            print("  [remount failed]", e)
        time.sleep(5)
    n_local = _count_hdf5(LOCAL_DATA_DIR)
    if errors or n_local < MIN_HDF5:
        raise RuntimeError(
            f"Copy incomplete after {MAX_PASSES} passes: {n_local} local hdf5, "
            f"{len(errors)} remaining errors (sample: {errors[:3]}). "
            "Re-run this cell — already-copied files are kept."
        )
    open(_marker, "w").close()
    print(f"Copy complete: {n_local} hdf5 files on local disk.")
else:
    print(f"Local dataset ready: {_count_hdf5(LOCAL_DATA_DIR)} hdf5 at {LOCAL_DATA_DIR}")

# Fresh start — no resume; write a trainer yaml with a clean ckpt dir
CKPT_DIR = "/content/drive/MyDrive/wildfire_runs/checkpoints_utae_fresh"
os.makedirs(CKPT_DIR, exist_ok=True)
trainer_yaml = "/tmp/trainer_utae_fresh.yaml"
with open(os.path.join(REPO_DIR, "cfgs/trainer_colab.yaml")) as f:
    trainer_cfg = yaml.safe_load(f)
for cb in trainer_cfg.get("callbacks", []):
    if "ModelCheckpoint" in str(cb.get("class_path", "")):
        cb.setdefault("init_args", {})["dirpath"] = CKPT_DIR
trainer_cfg["precision"] = "32-true"
with open(trainer_yaml, "w") as f:
    yaml.safe_dump(trainer_cfg, f, sort_keys=False)
print("Fresh UTAE start. Checkpoints:", CKPT_DIR)

EPOCHS = 100
BATCH_SIZE = 8

%cd {REPO_DIR}/src
!python train.py \
  --config=../cfgs/UTAE/all_features.yaml \
  --trainer={trainer_yaml} \
  --data=../cfgs/data_multitemporal_full_features.yaml \
  --seed_everything=0 \
  --trainer.max_epochs={EPOCHS} \
  --optimizer.init_args.lr=0.001 \
  --data.data_dir="{LOCAL_DATA_DIR}" \
  --data.batch_size={BATCH_SIZE} \
  --data.num_workers=2 \
  --do_test=True
```

## Notes & variations

- **Other models:** swap the `--config` file, e.g. `../cfgs/convlstm/full_run.yaml`, `../cfgs/UTAE/all_features.yaml`, or `../cfgs/LogisticRegression/full_run.yaml`. **Temporal models (UTAE, ConvLSTM) also require the multitemporal data config** `../cfgs/data_multitemporal_full_features.yaml` instead of the monotemporal one — UTAE additionally needs `return_doy: true`, which that config sets. See section 9 for a ready-to-run UTAE cell.
- **Different fold / test year:** add `--data.data_fold_id=N` (0–11). Fold 0 uses test year 2021.
- **Quick smoke test** (verify the pipeline end-to-end fast): add `--trainer.limit_train_batches=2 --trainer.limit_val_batches=2 --trainer.limit_test_batches=2 --trainer.max_epochs=1`.
- **Test only from a checkpoint:** add `--do_train=False --ckpt_path=/path/to/ckpt.ckpt`.
- Checkpoints/logs are written under `src/lightning_logs/` and (with `log_model: true`) uploaded to W&B.
